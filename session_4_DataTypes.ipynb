**octa number system**

- it is denoted with 0o or 0o
- octa means 8
- we have generally o to 9 digits available
- octa means only two digits allowed : 0 1 2 3 4 5 6 7  
- ex: 0o567 0o345
- not valid : 0o123 , 0o12345

```python
0o67
```

```python
0o1234567
```

**hexa number system**

- it is denoted with 0x or 0x
- hexa means 16
- we have generally o to 9 digits available
